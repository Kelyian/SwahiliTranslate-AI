```python
#loading necessary libraries
import pandas as pd

```

```python
#loading the training dataset 
train_df = pd.read_csv(
    r"C:\Users\sanke\Desktop\SwahiliTranslate AI\data\raw\train.tsv",
    sep="\t",
    header=None,
    names=["en", "swa"],
    skiprows=1
)

print(train_df.head())
print(train_df.columns)
#shape of the dataset
print("Shape:", train_df.shape)
```

```python
#checking for empty or whitespace-only English sentences 
print("Empty English sentences:", (train_df["en"].str.strip() == "").sum())

#checking for empty or whitespace-only Kiswahili sentences
print("Empty Kiswahili:", (train_df["swa"].str.strip() == "").sum())
```

```python
#removing duplicate rows
train_clean = train_df.drop_duplicates()
print("Shape after removing duplicates:", train_clean.shape)
```

```python
#checking remaining duplicates
print("Remanining duplicates:", train_clean.duplicated().sum())

#missing values
print("Missing values:\n", train_clean.isnull().sum())
```

```python
#investigating unusual long sentences
# Find sentences longer than 100 words
long_sentences = train_clean[
    (train_clean["en"].apply(lambda x: len(x.split())) > 100) |
    (train_clean["swa"].apply(lambda x: len(x.split())) > 100)
]

print("Number of unusually long pairs:", len(long_sentences))

print(long_sentences.head())
```

```python
#saving the cleaned dataset to a new file
train_clean.to_csv(r"C:\Users\sanke\Desktop\SwahiliTranslate AI\data\processed\train_cleaned.tsv", 
                   sep="\t", 
                   index=False)

print("Clean training dataset saved")
```

```python
dev = pd.read_csv(
    r"C:\Users\sanke\Desktop\SwahiliTranslate AI\data\raw\dev.tsv",
    sep="\t",
    header=None,
    names=["en", "swa"],
    skiprows=1
)

test = pd.read_csv(
    r"C:\Users\sanke\Desktop\SwahiliTranslate AI\data\raw\test.tsv",
    sep="\t",
    header=None,
    names=["en", "swa"],
    skiprows=1
)
# Check their sizes
print("Validation:", dev.shape)
print("Test:", test.shape)
```

```python
print("Validation missing values:")
print(dev.isnull().sum())

print("Validation duplicates:", dev.duplicated().sum())

print("\nTest missing values:")
print(test.isnull().sum())

print("Test duplicates:", test.duplicated().sum())
```

```python
# Remove duplicate translation pairs
dev_clean = dev.drop_duplicates()
test_clean = test.drop_duplicates()

# Check new sizes
print("Validation:", dev_clean.shape)
print("Test:", test_clean.shape)
```

```python
# Save cleaned validation dataset
dev_clean.to_csv(
    r"C:\Users\sanke\Desktop\SwahiliTranslate AI\data\processed\dev_clean.tsv",
    sep="\t",
    index=False
)

# Save cleaned test dataset
test_clean.to_csv(
    r"C:\Users\sanke\Desktop\SwahiliTranslate AI\data\processed\test_clean.tsv",
    sep="\t",
    index=False
)

print("Validation and test datasets saved!")
```
